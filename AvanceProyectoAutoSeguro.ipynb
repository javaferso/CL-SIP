```python
# --------------------------------------------
# Programa: Gestión de Clientes – Auto Seguro S.A.
# Funcionalidades implementadas:
#   1. Registrar cliente
#   2. Consultar cliente por RUT
#   3. Listar todos los clientes
#
# Profesor: Luis Tapia Aedo
# Asignatura: Introducción a la Programación
# Fecha: 19 Abril 2025
# --------------------------------------------

def validar_rut(rut: str) -> bool:
    """
    Verifica que el RUT tenga formato válido:
      - Un solo guión separando cuerpo y dígito verificador
      - Cuerpo de 7 u 8 dígitos
      - Dígito verificador: un dígito o 'K'
    Retorna True si el formato es correcto, False en caso contrario.
    """
    partes = rut.split('-')
    if len(partes) != 2:
        return False
    cuerpo, dv = partes
    if not (cuerpo.isdigit() and 7 <= len(cuerpo) <= 8):
        return False
    dv = dv.upper()
    return (dv.isdigit() and len(dv) == 1) or dv == 'K'


def validar_telefono(telefono: str) -> bool:
    """
    Verifica que el teléfono contenga sólo dígitos
    y tenga entre 7 y 9 caracteres.
    """
    return telefono.isdigit() and 7 <= len(telefono) <= 9


def registrar_cliente(clientes: dict):
    """
    Solicita datos de un cliente, valida RUT y teléfono,
    y almacena el registro en el diccionario 'clientes'.
    """
    print("\n--- Registrar Cliente ---")
    rut = input("Ingrese RUT (ej: 12345678-5): ").strip()
    while not validar_rut(rut):
        print("Error: formato de RUT inválido.")
        rut = input("Reingrese RUT (ej: 12345678-5): ").strip()
    if rut in clientes:
        print("Error: ese RUT ya está registrado.")
        return

    nombre = input("Ingrese nombre completo: ").strip()
    direccion = input("Ingrese dirección: ").strip()

    telefono = input("Ingrese teléfono (solo dígitos): ").strip()
    while not validar_telefono(telefono):
        print("Error: teléfono inválido.")
        telefono = input("Reingrese teléfono (7–12 dígitos): ").strip()

    correo = input("Ingrese correo electrónico: ").strip()

    clientes[rut] = {
        "Nombre": nombre,
        "Dirección": direccion,
        "Teléfono": telefono,
        "Correo": correo
    }
    print("Cliente registrado exitosamente.")


def consultar_por_rut(clientes: dict):
    """
    Solicita un RUT y muestra los datos del cliente si existe.
    """
    print("\n--- Consultar Cliente por RUT ---")
    rut = input("Ingrese RUT a consultar: ").strip()
    datos = clientes.get(rut)
    if datos:
        print(f"\nDatos del cliente con RUT {rut}:")
        print(f"  Nombre:    {datos['Nombre']}")
        print(f"  Dirección: {datos['Dirección']}")
        print(f"  Teléfono:  {datos['Teléfono']}")
        print(f"  Correo:    {datos['Correo']}\n")
    else:
        print("No se encontró cliente con ese RUT.\n")


def listar_clientes(clientes: dict):
    """
    Muestra un listado numerado de todos los clientes registrados.
    """
    print("\n--- Listar Todos los Clientes ---")
    if not clientes:
        print("No hay clientes registrados.\n")
        return

    for idx, (rut, datos) in enumerate(clientes.items(), start=1):
        print(f"{idx}. RUT: {rut}")
        print(f"   Nombre:    {datos['Nombre']}")
        print(f"   Dirección: {datos['Dirección']}")
        print(f"   Teléfono:  {datos['Teléfono']}")
        print(f"   Correo:    {datos['Correo']}\n")


def menu():
    """
    Menú principal:
      1. Registrar cliente
      2. Consultar cliente por RUT
      3. Listar todos los clientes
      4. Salir
    """
    clientes = {}
    while True:
        print("=== Menú de Gestión de Clientes ===")
        print("1. Registrar cliente")
        print("2. Consultar cliente por RUT")
        print("3. Listar todos los clientes")
        print("4. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            registrar_cliente(clientes)
        elif opcion == "2":
            consultar_por_rut(clientes)
        elif opcion == "3":
            listar_clientes(clientes)
        elif opcion == "4":
            print("Fin del programa.")
            break
        else:
            print("Opción no válida, intente de nuevo.\n")


if __name__ == "__main__":
    menu()
```
